```python
# apliquer le dict
SAFE_LEXICON = {
    ("cnx", "cnxn", "cnxx", "conex", "conx", "conixion"): "connexion",
    ("srvr", "servr", "server"): "serveur",
    ("optc", "optic"): "optique",
    ("net", "inter"): "internet",
    ("modm",): "modem",
    ("app", "appli", "apl"): "application",
    ("wi-fi", "wi fi", "wfi"): "wifi",
    ("rteur", "rtr"): "routeur",
    ("dsl", "adsl", "أدياسال", "اديسل"): "adsl",
    ("fiber", "fibr", "lfibre"): "fibre",
    ("num", "nmr", "nm"): "numero",
    ("idome", "adoom", "doom"): "idoom",
    ("هيوط",): "خيوط",
    ("جيڨابايت",): "جيقا",
    ("لاريسو",): "الريزو",
    ("Gp",): "Gbps"
}

def clean_text(text):
    if not isinstance(text, str):
        return text

    for errors, correct in SAFE_LEXICON.items():
        for err in errors:
            # remplacer mot par mot (safe)
            text = re.sub(r'\b' + re.escape(err.lower()) + r'\b', correct, text)

    return text

# appliquer sur la colonne texte
df["Commentaire client"] = df["Commentaire client"].apply(clean_text)

# sauvegarder nouveau fichier
df.to_csv("DATA\MergedComplaints.csv", index=False, encoding="utf-8")
```

```python
import pandas as pd
import re

# 1. Ta liste de référence (Base de noms arabes)
noms_reference = {
    "Rayan", "Amine", "Sami", "Yasmine", "Karim", "Zineb", "Omar", "Leila", 
    "Ahmed", "Mohamed", "Fatima", "Ali", "Hassan", "Idriss", "Youssef", "Anis",
    "Sofiane", "Myriam", "Sarah", "Hamza", "Walid", "Abdellah", "Meriem" , "raouf"
}

df=pandas.read_csv('DATA\MergedComplaints.csv')

def trouver_noms(phrase):
    # Cherche les mots commençant par une Majuscule (Noms propres potentiels)
    mots_majuscule = re.findall(r'\b[A-Z][a-z]+\b', phrase)
    
    # Filtre ceux qui sont dans notre liste de noms arabes
    trouves = [nom for nom in mots_majuscule if nom in noms_reference]
    return ", ".join(trouves) if trouves else "Aucun"

# 3. Application ligne par ligne
df['noms_arabes_detectes'] = df["Commentaire client"].apply(trouver_noms)

df['noms_arabes_detectes'].value_counts()
```
